# Section 6: Limitations, Policy Implications, and Extensions

The analysis describes conditional associations. Data timing, measurement, sample selection and model assumptions limit causal and policy conclusions.


<a id="section-6a"></a>

## 6(a) Key Data, Methodological, and Scope Limitations


The monthly outcome assumes recorded states persist until the next transaction over retained intervals, excludes UNKNOWN starts and gaps above 24 hours, and uses ≥25 raw-active days rather than verified full monitoring. Sensor selection, systematic reporting (including Sundays), unequal retained minutes and unverified boundary coverage limit outcome validity. Inventory time-limit mapping (including `2HR-30MIN`) and broader coordinate anomalies remain source caveats.

August 2025 occupancy is paired with October 2026 price/business snapshots, 2020–2024 ACS and 2024 geography. Business registrations are not necessarily distinct storefronts active during the outcome month, and missing coordinates/classifications limit counts. Only 49/4,139 zoning codes match; zoning cannot reliably control land use, and no nearest-parcel substitute is invented.

OLS is unconstrained for a bounded outcome. Residual diagnostics reject constant variance and the tested linear functional form; these findings qualify inference. Multiple spaces share only 45 common-sample tracts; repeated neighborhood values and local shocks can correlate errors. HC1, row bootstrap and random-space CV do not eliminate that dependence. Selection-conditioned p-values are not adjustment for the search across candidate models. These issues qualify inference even when slopes are significant.

<a id="section-6b"></a>

## 6(b) Potential Sources of Bias


Demand-responsive tariff setting and omitted employment, parking supply, enforcement, destinations and events can jointly affect prices and occupancy; this can generate a positive association despite downward demand. Later-vintage proxies introduce measurement error that need not be classical or simply attenuating. Sensor-equipped spaces, inventory linkage and common-sample exclusion can select the population. Business composition and residential mode shares are imperfect proxies for actual parker demand. No net direction for all biases can be identified from these data; reversing a coefficient would not resolve them.

<a id="section-6c"></a>

## 6(c) Policy Implications


The findings motivate collecting historically aligned prices, occupancy and supply before evaluating demand-responsive pricing. They do not justify raising tariffs because a cross-sectional slope is positive, nor removing commercial activity because a conditional composition slope is negative. Assess availability, turnover, spillovers, equity and business access jointly. Monthly average occupancy cannot set a live parking recommendation or a universal occupancy target for every neighborhood.

<a id="section-6d"></a>

## 6(d) Concrete Future Research Extensions


First, rebuild separate outcomes from original transactions under 12/24/48-hour gaps and 20/25/28-day eligibility, and compare Sundays with other days. Separate a common-space outcome change from expanded-sample composition and recheck predictor linkage. Second, align tariff/business vintages with the outcome month and collect curb supply/enforcement controls. Third, use repeated tariff/occupancy periods with a credible identification design to examine demand response; report pretrends and confounding rather than equating time changes with causality. Fourth, evaluate spatially grouped and future-time held-out samples and tract-aware inference. These are proposed extensions, not methods claimed to have been executed.

<a id="section-6e"></a>

## 6(e) External Validity


Evidence concerns retained LADOT sensor-equipped metered spaces represented in August 2025 and the observed tract contexts. It does not establish transfer to all LA neighborhoods, unmetered/private/off-street parking, other cities, future tariff regimes or time-specific availability. Random five-fold results test other spaces from the same observed setting; they are not a transfer test. The historically mismatched proxies further restrict interpreting the fitted associations as an exact snapshot of August 2025 demand.
