# Section 6: Limitations, Policy Implications, and Extensions

Model 3 estimates a positive conditional association between the stated hourly price proxy and monthly occupancy. A $1 higher price proxy is associated with about 5.15 percentage points higher occupancy, holding the included variables constant. This does not show that raising a meter price would increase occupancy. The model explains 29.4% of the observed variation, and its mean five-fold prediction error is about 16.3 percentage points. These results need to be read in light of how the data were measured and combined, how Model 3 was selected, and the limits of the diagnostic and validation methods.

<a id="section-6a"></a>

## 6(a) Limitations

The outcome is a monthly occupancy fraction constructed from sensor transactions. The calculation carries each recorded state forward until the next transaction, excludes intervals that begin with an UNKNOWN state and drops gaps longer than 24 hours. It also requires at least 25 raw-active days, which does not guarantee complete monitoring throughout the month. Differences in sensor coverage, reporting patterns, retained minutes and tract boundaries can therefore affect the measured occupancy rate. The inventory’s time-limit categories and coordinate anomalies add uncertainty to some space characteristics. The outcome is a measure of occupancy during retained intervals, not a live measure of whether a driver can find a space.

The data sources also refer to different periods. August 2025 occupancy is matched with October 2026 price and business snapshots, 2020–2024 ACS estimates and 2024 geography. The later price and business records may not describe conditions when occupancy was measured. Business registrations do not necessarily represent distinct storefronts operating during August 2025, and missing coordinates and classifications limit the counts. Only 49 of 4,139 zoning codes matched; zoning therefore provides limited control for local land use.

The model has further limits. Occupancy is bounded between zero and one, while OLS does not constrain fitted values to that range. The residual diagnostics indicate unequal error variance and reject the tested linear functional form. HC1 adjusts standard errors for heteroskedasticity under independent observations, but it does not address misspecification or dependence among spaces in the same tract. The 4,016 observations come from only 45 common-sample tracts, so shared neighborhood conditions may make errors correlated. Row bootstrap intervals and random-space cross-validation also treat spaces as independent; they do not remove that concern. Finally, Model 3 was selected after screening candidate predictors. Its reported p-values and intervals describe the selected model and do not adjust for the full search across candidate specifications.

<a id="section-6b"></a>

## 6(b) Why the estimated relationships may differ from demand effects

The positive price association may reflect how meter rates are set as well as how drivers respond to price. If higher-demand areas receive higher rates, price and occupancy can move together across spaces even when an individual driver is less likely to park as the price rises. Employment, parking supply, enforcement, destinations and events may also affect both rates and occupancy, but are not fully measured here. The cross-sectional model cannot separate these influences from a causal response to price.

The business and travel variables are also proxies for local conditions. In Model 3, both total business registrations and food-service/retail registrations are included. The food-service/retail coefficient therefore describes a change in business composition at a given total count; it is not the effect of opening more retail businesses. Transit and walking shares describe tract residents, not necessarily the people who parked at each meter. Differences in registration coverage, timing and sensor inclusion may introduce additional bias. The available data do not establish a single direction for all these sources of bias.

<a id="section-6c"></a>

## 6(c) Policy implications

The positive price coefficient should not be used by itself to recommend raising meter rates. It describes differences across spaces in a cross-section, where rates may already reflect demand and neighborhood conditions. Likewise, the negative conditional association for food-service/retail registrations is not evidence that commercial activity should be reduced. The results are useful for identifying patterns to investigate, but do not estimate the effect of changing prices or land use.

Parking policy should consider occupancy together with availability, turnover, spillovers to nearby blocks, equity and access for local businesses. A monthly average cannot guide real-time pricing or establish one occupancy target for every neighborhood. Evaluating rate changes requires prices and occupancy measured over aligned periods, along with information about supply and other management policies.

<a id="section-6d"></a>

## 6(d) Future research

A next study could first rebuild occupancy from the original transactions using alternative gap and active-day rules, including 12-, 24- and 48-hour gaps; 20-, 25- and 28-day thresholds; and comparisons of Sundays with other days. It should distinguish changes in occupancy for the same spaces from changes caused by adding different spaces, and verify sensor coverage and predictor linkage.

The analysis could then align prices and business activity with the occupancy month and add measures of curb supply, enforcement and local demand. Better parcel or zoning matches would improve the description of land use. To estimate how drivers respond to price, a study would need repeated price and occupancy observations and a credible design for separating rate changes from underlying demand; time trends alone would not establish causality. Future validation could hold out entire tracts or later time periods and use inference that accounts for spaces sharing tract conditions. These are proposed extensions; they were not carried out in the current analysis.

<a id="section-6e"></a>

## 6(e) Scope of the findings

The findings apply to the retained LADOT sensor-equipped metered spaces observed in August 2025 and to the neighborhood contexts represented by the 45 common-sample tracts. They do not establish what occupancy is like at every Los Angeles space, at unmetered, private or off-street parking, in other cities, or under future rate schedules. The five-fold results measure prediction error for held-out spaces from the same observed setting. Because spaces from shared tracts can appear in both training and test folds, those results do not show how the model performs in entirely new neighborhoods or future periods. The timing differences between occupancy and several predictors further limit treating the estimates as a precise account of August 2025 conditions.
