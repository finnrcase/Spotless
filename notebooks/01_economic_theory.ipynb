## 1(b) Relevant Economic Theory and Academic Literature

**Requirement:** Review theory in 2–3 paragraphs and cite at least two academic sources.

**Method:** Review the two cited academic studies in three theory paragraphs. **Evidence:** current code/outputs or cited substantive discussion immediately below. **Interpretation:** accompanies the evidence, including units and limitations. **Status:** Complete for analytical coverage; execution evidence is recorded separately, and no grade is guaranteed.

Shoup (2006) models the choice between searching for a cheaper curb space and paying for immediately available off-street parking. The relative monetary and time costs explain why scarce curb space can produce cruising. This motivates occupancy as an economically relevant outcome and a downward demand response to price, holding other demand conditions constant. We do not measure cruising directly. [Shoup, *Transport Policy* 13(6), 479–486](https://escholarship.org/content/qt55s7079f/qt55s7079f.pdf).

Pierce and Shoup (2013) study SFpark price/occupancy adjustments and report average price elasticity about −0.4, with substantial heterogeneity. Their setting illustrates why tariff changes and location differences answer different questions: administrators can set higher tariffs where demand is stronger. [Pierce and Shoup, *JAPA* 79(1), 67–81, doi:10.1080/01944363.2013.787307](https://www.its.ucla.edu/wp-content/uploads/sites/10/2017/06/Getting-the-Prices-Right-An-Evaluation-of-Pricing-Parking-by-Demand-in-San-Francisco.pdf).

In our cross-section, nearby activity and residential density can shift parking demand, while walking/transit alternatives and local parking supply can offset it. Resident commuting shares are contextual proxies, not the modes of the actual parkers. Therefore price endogeneity, destination sorting and omitted supply prevent interpreting an OLS coefficient as a structural demand response. We estimate conditional associations, and compare them with the demand hypothesis without claiming causal identification.

## 1(d) Expected Coefficient Signs and Magnitudes

**Requirement:** Discuss both expected signs and magnitudes in economic units.

**Method:** Translate a literature elasticity into an illustrative magnitude and discuss every candidate’s scale. **Evidence:** current code/outputs or cited substantive discussion immediately below. **Interpretation:** accompanies the evidence, including units and limitations. **Status:** Complete for analytical coverage; execution evidence is recorded separately, and no grade is guaranteed.

The demand hypothesis predicts a negative **causal** price response conditional on demand; observed cross-location price associations can have either sign when prices respond to demand. A useful magnitude benchmark is elasticity ε≈−0.4 from [Pierce and Shoup (2013)](https://www.its.ucla.edu/wp-content/uploads/sites/10/2017/06/Getting-the-Prices-Right-An-Evaluation-of-Pricing-Parking-by-Demand-in-San-Francisco.pdf). At an **illustrative** 50% occupancy and \$2/hour price, a 10% price rise (\$0.20) implies approximately a 4% relative occupancy decline, or **2 occupancy percentage points**. Locally, β≈εq/p≈−0.10 occupancy-fraction units per dollar. This is a transportability benchmark, not an imposed coefficient, a preregistered prior, or a causal estimate for Los Angeles. Large \$1 changes need not follow the local approximation.

The other slopes have no calibrated structural magnitudes in these two sources. Discussing size therefore requires units, scale and the occupancy bound rather than invented numerical priors:

| Candidate | Demand-based sign expectation / qualification | Magnitude scale |
|---|---|---|
| Price midpoint | Negative holding demand fixed; cross-location sign can reverse | \$0.20 and \$1 changes; benchmark above; not an elasticity coefficient |
| Time limit | Longer stays can raise occupancy; turnover response makes sign ambiguous | 1 additional hour implies 100β percentage points |
| Population/housing density | Positive demand pressure, offset by local supply/mode choice | 1,000 people or housing units per square mile implies 100,000β pp |
| Median income | Ambiguous: car demand versus price/time sensitivity | \$10,000 implies 1,000,000β pp; no calibrated prior |
| Food-service/retail and all-business counts | Positive demand pressure in isolation; conditional composition effects can reverse signs | 100 registrations implies 10,000β pp; not 100 distinct new storefronts |
| Drive-alone / transit / walk shares | Ambiguous resident demand, substitution and centrality mechanisms | A 10-pp share change implies 10β occupancy pp; shares are fractions |
| TOD / meter technology indicators | Ambiguous assignment and location differences | Indicator switch implies 100β pp relative to reference |

Effects are local comparisons within observed support, not unlimited linear extrapolations. Occupancy is between zero and one: a 5-pp change is meaningful against 50% occupancy, but is a 10% **relative** change, not 5%. Boundedness, saturation and overlapping covariates limit plausible effect sizes. These expectations are stated explicitly for interpretation; estimates below are not modified to match them.

## 1. Research question and sample

How is metered parking occupancy associated with the stated price proxy and local demand conditions? I summarize Notebooks 01–07 using **4,016 parking spaces across 45 Census tracts**, with August 2025 time-weighted occupancy as the dependent variable. The primary specification remains the seven-variable Model 3.

Shoup (2006), *Cruising for parking*, connects curb prices, scarce spaces and the incentive to search for parking. Cheap curb parking can encourage cruising when spaces are occupied, so pricing matters for both availability and congestion. This motivates studying occupancy rather than counting transaction records. [Transport Policy, 13(6), 479–486](https://escholarship.org/content/qt55s7079f/qt55s7079f.pdf).

Pierce and Shoup (2013), *Getting the Prices Right: An Evaluation of Pricing Parking by Demand in San Francisco*, studies a program that adjusts parking prices with demand. That setting highlights why a comparison across locations can combine drivers’ response to price with officials’ response to demand. The Los Angeles coefficient below is a conditional association, not the effect of an externally imposed price change. [Journal of the American Planning Association, 79(1), 67–81](https://doi.org/10.1080/01944363.2013.787307).

My starting demand hypothesis is a negative price coefficient, holding demand conditions fixed. I expect density and nearby commercial activity to increase demand; longer time limits may raise occupancy through longer stays, while commuting shares and TOD status have ambiguous signs in this resident-based cross-section. Theory does not pin down a credible numerical magnitude here, so I report percentage-point differences and confidence intervals rather than invent an expected effect size.
